```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime
```

```python
dates = pd.date_range('20190214', periods=6)

numbers = np.matrix([[101, 103], [105.5, 75], [102, 80.3], [100, 85], [110, 98], [109.6, 125.7]])

df = pd.DataFrame(numbers, index=dates, columns=['A', 'B'])
```

```python
print(df.loc['2019-02-18'])
```

```python
print(df.loc[datetime(2019, 2, 18 )])
```

```python
print(df.iloc[-2])
```

```python
print(df.iloc[:2,  1:2])
```

```python
print(df.sort_values(by='B', ascending=False))
```

```python
print(df['A'].max())
```

```python
df.loc[df['A'].idxmax(), 'A'] *= 2
print(df)
```

```python
print(df[df['A'] > 105])
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df['A'], marker = 'o')
plt.title('A')
plt.xlabel('Date')
plt.ylabel('Value')
plt.grid
plt.show
```

```python
print(df[df['B']<df['A']])
```

```python
#4
a = np.random.randint(1, 6, 10)
print(a)

print(np.where(a[:-1] == a[1:])[0] + 1)
```

```python
#5
a = np.random.randint(low = -10, high= 10, size = 10)
b = np.random.randint(low = -10, high= 10, size = 10)

print(a)
print(b)

print(np.where(a>b)[0])
```

```python
#6
a = np.random.randint(1, 11, 10)
print(a)

a[:-1] = a[1:] 
a[-1] = a[-2]

print(a)
```

```python
#7
a = np.random.randint(1, 11, 10)

print(a)

a = a[::-1]

print(a)
```
